```python
def substitution_score(a, b):
    if a == '-' or b == '-':
        return 0 
    return BLOSUM62.get(a, {}).get(b, -4)
```

```python
def pairwise_affine_gap_penalty(seq1, seq2):
    GAP_OPEN = -10
    GAP_EXTEND = -1
    
    penalty = 0
    in_gap1, in_gap2 = False, False

    for a, b in zip(seq1, seq2):
        # Gap in seq1
        if a == '-' and b != '-':
            if not in_gap1:
                penalty += GAP_OPEN
                in_gap1 = True
            else:
                penalty += GAP_EXTEND
        else:
            in_gap1 = False

        # Gap in seq2
        if b == '-' and a != '-':
            if not in_gap2:
                penalty += GAP_OPEN
                in_gap2 = True
            else:
                penalty += GAP_EXTEND
        else:
            in_gap2 = False

        # No extra penalty if both are gaps
    return penalty
```

```python
def calculate_sp_score(msa):
    n_seq = len(msa)
    L = len(msa[0])
    total_score = 0.0

    for (i, j) in combinations(range(n_seq), 2):
        seq1, seq2 = msa[i], msa[j]
        pair_score = 0.0

        # Substitution scores
        for a, b in zip(seq1, seq2):
            if a == '-' and b == '-':
                continue  # no score for double gap
            pair_score += substitution_score(a, b)

        # Add affine gap penalty
        pair_score += pairwise_affine_gap_penalty(seq1, seq2)
        total_score += pair_score

    return total_score
```

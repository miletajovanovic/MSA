# Compare Brute Force vs Genetic Algorithm for MSA

```python
# Import both algorithms
%run brute_force.ipynb
%run ga.ipynb

import time
```

## Comparison Function

```python
def compare_algorithms(sequences, max_gaps=3, ga_params=None):

    if ga_params is None:
        ga_params = {
            'population_size': 500,
            'num_generations': 100,
            'tournament_size': 50,
            'mutation_prob': 0.1,
            'elitism_size': 25
        }
    
    print(f"Sequences: {sequences}")
    
    # Run Brute Force
    print("\nRunning Brute Force...")
    start = time.time()
    bf_alignment, bf_score = brute_force_msa(sequences, max_gaps=max_gaps)
    bf_time = time.time() - start
    
    print(f"Brute Force - Score: {bf_score:.2f}, Time: {bf_time:.3f}s")
    for seq in bf_alignment:
        print(f"  {seq}")
    
    # Run GA
    print("\nRunning Genetic Algorithm...")
    start = time.time()
    ga_individual, _ = ga_msa(sequences, max_gaps=max_gaps, **ga_params)
    ga_time = time.time() - start
    ga_score = ga_individual.fitness
    
    print(f"GA - Score: {ga_score:.2f}, Time: {ga_time:.3f}s")
    for seq in ga_individual.code:
        print(f"  {seq}")
    
    # Compare
    diff = bf_score - ga_score  # How much worse GA is (positive means GA is worse)
    is_optimal = abs(diff) < 0.01  # GA found optimal solution
    speedup = bf_time / ga_time if ga_time > 0 else 0
    
    print(f"\nComparison:")
    print(f"  Score Difference: {diff:.2f} (BF - GA)")
    print(f"  Optimal: {'YES' if is_optimal else 'NO'}")
    print(f"  Speedup: {speedup:.2f}x")
    print("="*60)
    
    return {
        'sequences': sequences,
        'bf_score': bf_score,
        'ga_score': ga_score,
        'bf_time': bf_time,
        'ga_time': ga_time,
        'diff': diff,
        'is_optimal': is_optimal,
        'speedup': speedup
    }
```

## Test Sequences

```python
# 20 test cases with varying complexity
test_sequences = [
    ['ACT', 'ACT', 'ACT'],
    ['ACT', 'AT', 'ACT'],
    ['AG', 'A', 'AG'],
    ['ACGT', 'AGT', 'AT'],
    ['GAT', 'AT', 'T'],
    ['GATTACA', 'GTACA', 'GACA'],
    ['ATCG', 'ACG', 'ATG'],
    ['CGTA', 'CTA', 'CGA'],
    ['ACE', 'AE', 'CE'],
    ['PAWHE', 'HEAWY', 'AWHE'],
    ['ACDEG', 'ACEG', 'ADEG'],
    ['PAWH', 'AWH', 'PWH', 'AH'],
    ['ACGT', 'AGT'],
    ['HELIX', 'HEAX'],
    ['A', 'C', 'T'],
    ['W', 'H', 'E'],
    ['ATGATG', 'ATGTG', 'ATG'],
    ['ABCD', 'BCD', 'CD', 'D']
]
```

## Run Comparisons

```python
# Run all comparisons
results = []

for i, seqs in enumerate(test_sequences):
    print(f"\n{'='*60}")
    print(f"Test {i+1}/{len(test_sequences)}")
    print(f"{'='*60}")
    
    result = compare_algorithms(seqs, max_gaps=3)
    results.append(result)
```

## Summary

```python
print("\n" + "="*150)
print("SUMMARY OF ALL TESTS")
print("="*150)
print(f"{'Test':<5} {'Sequences':<30} {'BF Score':>10} {'GA Score':>10} {'Diff':>8} {'Optimal':>10} {'BF Time':>10} {'GA Time':>10} {'Speedup':>10}")
print("-"*150)

for i, r in enumerate(results, 1):
    seqs_str = str(r['sequences'])[:30]
    optimal = 'YES' if r['is_optimal'] else 'NO'
    print(f"{i:<5} {seqs_str:<30} {r['bf_score']:>10.2f} {r['ga_score']:>10.2f} {r['diff']:>8.2f} {optimal:>10} {r['bf_time']:>9.3f}s {r['ga_time']:>9.3f}s {r['speedup']:>9.2f}x")
```
